```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.style
```

Đọc dữ liệu đầu vào từ file csv

```python
data_dir = Path(r"D:\AWS\Datasets\Marketing & E-Commerce Analytics Dataset")

# Chọn đúng ba file cần thiết
required_files = ['events.csv', 'products.csv', 'transactions.csv']
selected_files = [data_dir / fname for fname in required_files]

# Kiểm tra file tồn tại
missing_files = [f for f in selected_files if not f.exists()]
if missing_files:
    raise FileNotFoundError(f'Missing required CSV files: {missing_files}')

print('Các file CSV được chọn:')
for f in selected_files:
    print('-', f.name)

# Đọc từng file riêng lẻ
data_frames = {f.name: pd.read_csv(f) for f in selected_files}
for name, df in data_frames.items():
    print(f"{name}: {df.shape} rows, {df.shape[1]} columns")
```

Xem xét dữ liệu ban đầu

```python
for name, df in data_frames.items():
    print(f"--- {name} ---")
    display(df.head(20))
    print()
```

```python
for name, df in data_frames.items():
    print(f"--- {name} info ---")
    df.info()
    print()
```

Thống kê các chỉ số định lượng trên dữ liệu hiện tại

```python
for name, df in data_frames.items():
    print(f"--- Thống kê {name} ---")
    display(df.describe(include='all'))
    print()
```

Kiểm tra dữ liệu bị thiếu, dữ liệu bị trùng

```python
for name, df in data_frames.items():
    print(f"--- Kiểm tra {name} ---")
    print('Số lượng dòng bị trùng trong tập dữ liệu:', df.duplicated().sum())
    print('Số lượng giá trị bị thiếu trong toàn bộ DataFrame:', df.isna().sum().sum())
    print('Số lượng giá trị thiếu (NaN) theo cột:')
    print(df.isna().sum())
    print()
```

```python
for name, df in data_frames.items():
     print(f"{name} missing values:")
     display(df.isna().sum().to_frame('missing_count'))
```

```python
# Đọc các giá trị duy nhất của cột 'event_type', device_type, traffic_source, page_category, experiment_group trong events.csv
events_df = data_frames['events.csv']
unique_event_types = events_df['event_type'].unique()
unique_device_types = events_df['device_type'].unique()
unique_traffic_sources = events_df['traffic_source'].unique()
unique_page_categories = events_df['page_category'].unique()
unique_experiment_groups = events_df['experiment_group'].unique()
print("'event_type':", unique_event_types)
print("'device_type':", unique_device_types)
print("'traffic_source':", unique_traffic_sources)
print("'page_category':", unique_page_categories)
print("'experiment_group':", unique_experiment_groups)



# Đếm số lượng mỗi loại event_type, device_type, traffic_source, page_category, experiment_group trong events.csv
event_type_counts = events_df['event_type'].value_counts()
device_type_counts = events_df['device_type'].value_counts()
traffic_source_counts = events_df['traffic_source'].value_counts()
page_category_counts = events_df['page_category'].value_counts()
experiment_group_counts = events_df['experiment_group'].value_counts()

print("Số lượng mỗi loại 'event_type':")
print(event_type_counts)
print("Số lượng mỗi loại 'device_type':")
print(device_type_counts)
print("Số lượng mỗi loại 'traffic_source':")
print(traffic_source_counts)
print("Số lượng mỗi loại 'page_category':")
print(page_category_counts)
print("Số lượng mỗi loại 'experiment_group':")
print(experiment_group_counts)
```

```python
# Đọc các giá trị duy nhất của cột category trong products.csv
products_df = data_frames['products.csv']
unique_categories = products_df['category'].unique()
print("'category':", unique_categories)

# Đếm số lượng mỗi loại category trong products.csv
category_counts = products_df['category'].value_counts()
print("Số lượng mỗi loại 'category':")
print(category_counts)
```

```python
# đọc các giá trị duy nhất của cột quantity, discount_applied, refund_flag trong transactions.csv
transactions_df = data_frames['transactions.csv']
unique_quantities = transactions_df['quantity'].unique()
unique_discounts = transactions_df['discount_applied'].unique()
unique_refunds = transactions_df['refund_flag'].unique()
print("'quantity':", unique_quantities)
print("'discount_applied':", unique_discounts)
print("'refund_flag':", unique_refunds)

# đếm số lượng mỗi loại quantity, discount_applied, refund_flag trong transactions.csv
quantity_counts = transactions_df['quantity'].value_counts()
discount_counts = transactions_df['discount_applied'].value_counts()
refund_counts = transactions_df['refund_flag'].value_counts()
print("Số lượng mỗi loại 'quantity':")
print(quantity_counts)
print("Số lượng mỗi loại 'discount_applied':")
print(discount_counts)
print("Số lượng mỗi loại 'refund_flag':")
print(refund_counts)

# đếm số lượng gross_revenue bị âm trong transactions.csv
negative_revenue_count = (transactions_df['gross_revenue'] < 0).sum()
print("Số lượng gross_revenue bị âm:", negative_revenue_count)
```

```python
#in ra những dòng có refund_flag = 1, sắp xếp theo gross_revenue tăng dần
refund_transactions = transactions_df[transactions_df['refund_flag'] == 1].sort_values(by='gross_revenue')
print("Các giao dịch có refund_flag = 1, sắp xếp theo gross_revenue tăng dần:")
display(refund_transactions)
```

```python
# chuyển tên cột events.timestamp → event_timestamp, transactions.timestamp → transaction_timestamp, transactions.refund_flag → is_refunded
events_df.rename(columns={'timestamp': 'event_timestamp'}, inplace=True)
transactions_df.rename(columns={'timestamp': 'transaction_timestamp', 'refund_flag': 'is_refunded'}, inplace=True)
```

```python
#liệt kê tên các cột trong các file event, product, transaction
print("Các cột trong events.csv:", events_df.columns.tolist())
print("Các cột trong products.csv:", products_df.columns.tolist())
print("Các cột trong transactions.csv:", transactions_df.columns.tolist())
```

```python
# Chỉ dùng cột parsed để kiểm tra lỗi
invalid_event_timestamps = events_df["event_timestamp_parsed"].isna().sum()
print("Invalid event timestamps:", invalid_event_timestamps)
```

```python
# Vì không có lỗi, xóa cột tạm
events_df = events_df.drop(columns=["event_timestamp_parsed"])
```

```python
# Parse transaction_timestamp để kiểm tra lỗi
transactions_df["transaction_timestamp_parsed"] = pd.to_datetime(
    transactions_df["transaction_timestamp"],
    errors="coerce"
)

# Đếm số dòng lỗi datetime
invalid_transaction_timestamps = transactions_df["transaction_timestamp_parsed"].isna().sum()
print("Invalid transaction timestamps:", invalid_transaction_timestamps)

# Kiểm tra dtype
print(transactions_df[["transaction_timestamp", "transaction_timestamp_parsed"]].dtypes)
```

```python
transactions_df = transactions_df.drop(columns=["transaction_timestamp_parsed"])
```

```python
# Parse launch_date để kiểm tra lỗi format
products_df["launch_date_parsed"] = pd.to_datetime(
    products_df["launch_date"],
    errors="coerce"
)

# Đếm số ngày lỗi
invalid_launch_dates = products_df["launch_date_parsed"].isna().sum()
print("Invalid launch dates:", invalid_launch_dates)

# Kiểm tra dtype
print(products_df[["launch_date", "launch_date_parsed"]].dtypes)

# Nếu invalid = 0, xóa cột tạm
products_df = products_df.drop(columns=["launch_date_parsed"])
```

```python
print("events_df columns:")
print(events_df.columns.tolist())

print("\nproducts_df columns:")
print(products_df.columns.tolist())

print("\ntransactions_df columns:")
print(transactions_df.columns.tolist())
```

```python
from pathlib import Path

output_dir = Path("data/local_preprocessed")
output_dir.mkdir(parents=True, exist_ok=True)

events_df.to_csv(output_dir / "events.csv", index=False)
products_df.to_csv(output_dir / "products.csv", index=False)
transactions_df.to_csv(output_dir / "transactions.csv", index=False)

print("Export completed:")
print(output_dir / "events.csv")
print(output_dir / "products.csv")
print(output_dir / "transactions.csv")
```
